```python
print("Chaitanya M Mitra, 1WA24CS090")

def unify(x, y, substitution=None):
    if substitution is None:
        substitution = {}

    if x == y:
        return substitution

    # If x is a variable
    if isinstance(x, str) and x.islower():
        substitution[x] = y
        return substitution

    # If y is a variable
    if isinstance(y, str) and y.islower():
        substitution[y] = x
        return substitution

    return None


# Given expressions
expression1 = ["Eats", "x", "Apple"]
expression2 = ["Eats", "Riya", "y"]

# Check predicate symbols
if expression1[0] != expression2[0]:
    print("Unification Failed: Predicate symbols are different")

elif len(expression1) != len(expression2):
    print("Unification Failed: Number of arguments are different")

else:
    substitution = {}

    for i in range(1, len(expression1)):
        substitution = unify(
            expression1[i],
            expression2[i],
            substitution
        )

        if substitution is None:
            break

    if substitution is None:
        print("Unification Failed")
    else:
        print("Unification Successful!")
        print("Substitution:", substitution)
        print("Unified Expression: Eats(Riya, Apple)")
```
